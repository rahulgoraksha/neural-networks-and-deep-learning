Deep Learning : Deep Learning is a type of Machine Learning that uses many layered neural networks to learn patterns from data and make predictions or decisions.


A neural network is a computer model that is inspired by human brain that learns patterns from the data.


Each neuron does a small calucation or stores some memory.


Many neurons combines together to form or understand complex pattern.




**Types of Layers**

- Input Layer : This is where we pass the data to the neural network (Carpet area, no of rooms, no of bathroom)
  any fact or feature should be passed into the input layer.
  
  
- Hidden Layer : These are the middle layers of neural network. They are called hidden because we do not directly give answers there and we do not see the final result.


- Output Layer: This is the final answer produced by the neural network.

Limitation of Deep Learning:

- It needs a large volume of data.
- Neeed high computing power
- Black box Problem  (Sometimes model gives us the right answer, but it is difficult to understand why)
- Can lead to Overfit (performs well on the known data,but not on unseen data)



Types of Deep Learning Networks:

- FeedForward Neural Network : basic neural network (It is used for simple prediction and classification task)
- CNN (Convolutional Neural Network) : Is typically used for images and videos  (Face recg, object detection)
- RNN (Recurrent Neural Network) : It is used in sequence data (Time series, Speech Recognition, Text)
- Transformers : Used in Ai Models and LLM (Chatbot, Summary, Text generation)

# BackPropogation - How Neural Network Works

- Forward Propogation : it is the process of computing the output of a neural network by sequentially passing the input data through each layer (from input layer to output layer) using weights 


- Back Propogation : it is algorithm used to compute the gradient of the loss function of each weight 



* Forward Pass - Making Prediction 

* Backward Pass - finding errors

* Gradient Descent - Fixing weights

* Weights Updates - Improvise prediction
