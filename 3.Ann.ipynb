ANN is a ML model insprired by the way human brain processes information. It is made of small mathematical units called as neurons. These neurons receive inputs, processes it and pass the output to the next layer.


It is a collection of connected layers. Each layer contains neurons. Each neuron performs weighted calculation, adds a bias value, applies activation function and sends the results forward.


Input Layer --> Hidden Layers (s). ---> Output Layer

z = w1x1 + w2x2 + w3x3 + ..... b


where

x : input features

w : weight

b : bias

z : weighted sum (aactivation) output

**Why do we need ANN**

Traditional ML.                            ANN / Deep Learning


often needs manual feature engineering     can learn useful representation automatically

works well for structured data.            works well for structured and complex data

faster in smaller dataset.                 can become powerful with more data and computation

LR, DT, RF , SVM                           

ANN Architecture 


1. Input Layer : layer recieves the features from the datset

2. hidden layer : hidden layers are the learning area of the neural network

3. output layer : output layer gives the final outcome




Activation Function

ReLU.     0 to infinity 

Sigmoid   0 to 1

Softmax.  Multiclass output 



Loss Function

The loss function tells the model how wrong its prediction is . Lower the value means better prediction.


Binary classification : binary cross-entrophy

multi class : categorical cross-entrophy

regression : MAE,MSE,RMSE, R2



Optimizer

SGD.  : Basic gradient optimizer
Adam. : fast, stable,and commonly used otpimizer
RMSprop : used in some complex deep learing task




Epoch , Batch Size, Validation 

Epoch  : one full pass over the training dataset  (if epoch is 50 , model sees full training data 50 times)

Batch Size: Number of records processed at a time (batch size=32, the model processes 32 rows at a time)


Validation data : Data used during training for monitoring (helps detect the model is overfit)
